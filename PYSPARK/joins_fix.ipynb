```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder \
    .appName("PySpark_Practice") \
    .master("local[*]") \
    .getOrCreate()
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
dataj1 = [('1','gaur','d01'),
          ('2','kit','d02'),
          ('3','sam','d03'),
          ('4','tim','d03'),
          ('5','aman','d05'),
          ('6','nad','d06')] 

schemaj1 = 'emp_id STRING, emp_name STRING, dept_id STRING' 

df1 = spark.createDataFrame(dataj1,schemaj1)

dataj2 = [('d01','HR'),
          ('d02','Marketing'),
          ('d03','Accounts'),
          ('d04','IT'),
          ('d05','Finance')]

schemaj2 = 'dept_id STRING, department STRING'

df2 = spark.createDataFrame(dataj2,schemaj2)
```

```python
df1.show()
df2.show()
```

```python
df1.join(df2,df1['dept_id']==df2['dept_id'],'inner').show()
```

```python
df1.join(df2,df1['dept_id']==df2['dept_id'],'left').show()
```

```python
df1.join(df2,df1['dept_id']==df2['dept_id'],'right').show()
```

```python
df1.join(df2,df1['dept_id']==df2['dept_id'],'full').show()
```

```python
df1.join(df2,df1['dept_id']==df2['dept_id'],'anti').show()
```
