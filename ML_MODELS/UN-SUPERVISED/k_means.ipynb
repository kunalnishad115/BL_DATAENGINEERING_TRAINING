```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans 
```

```python
iris=load_iris()
```

```python
x=iris.data
```

```python
model=KMeans(
  n_clusters=3,random_state=42
)
model.fit(x)
```

```python
print(model.labels_)
print(model.cluster_centers_)
```

```python
plt.figure(figsize=(7,5))

plt.scatter(

    x[:,0],

    x[:,1],

    c=model.labels_,

    cmap="viridis"

)

plt.scatter(

    model.cluster_centers_[:,0],

    model.cluster_centers_[:,1],

    c="red",

    s=250,

    marker="X"

)

# plt.show()
```

```python
wcss = []

for i in range(1,11):

    model = KMeans(

        n_clusters=i,

        random_state=42

    )

    model.fit(x)

    wcss.append(model.inertia_)
```

```python
plt.plot(

    range(1,11),

    wcss,

    marker="o"

)

plt.xlabel("Number of Clusters")

plt.ylabel("WCSS")

plt.title("Elbow Method")

plt.show()
```
